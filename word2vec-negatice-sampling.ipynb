```python
import pandas as pd
import os

# Define the mapping for the labels
label_mapping = {
    'Offensive_Untargetede': 'Negative',
    'Offensive_Targeted_Insult_Group': 'Negative',
    'Offensive_Targeted_Insult_Individual': 'Negative',
    'Not_offensive': 'Neutral',
    'not-Tamil': '(Discard)'
}

# Define file paths
input_files = [
    '/home/nibiru/Documents/projects/Text_try2/DravidianCodeMix/tamil_offensive_full_train.csv',
]
output_dir = '/home/nibiru/Documents/projects/Text_try2/preprocessed_data'

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

for input_file in input_files:
    # Create output filename
    base_filename = os.path.basename(input_file)
    output_filename = base_filename.replace('.csv', '_preprocessed.csv')
    output_file = os.path.join(output_dir, output_filename)

    # Read the data
    # The file seems to be tab-separated with no header
    df = pd.read_csv(input_file, sep='\\t', header=None, names=['sentence', 'label'], engine='python', on_bad_lines='warn')

    # Apply the label mapping
    df['label'] = df['label'].map(label_mapping)

    # Filter out rows to be discarded
    df = df[df['label'] != '(Discard)']

    # Save the preprocessed data
    df.to_csv(output_file, index=False)

    print(f"Preprocessing complete for {base_filename}. Output saved to {output_file}")
    print(df.head())
    print("-" * 30)
```

```python
import pandas as pd
import os

# Define the mapping for the sentiment labels
sentiment_label_mapping = {
    'Positive': 'Positive',
    'Negative': 'Negative',
    'unknown_state': 'Neutral',
    'Mixed_feelings': 'Neutral',
    'not-Tamil': '(Discard)'
}

# Define file paths
input_file = '/home/nibiru/Documents/projects/Text_try2/DravidianCodeMix/tamil_sentiment_full_train.csv'
output_dir = '/home/nibiru/Documents/projects/Text_try2/preprocessed_data'
output_filename = 'tamil_sentiment_full_train_preprocessed.csv'
output_file = os.path.join(output_dir, output_filename)

# The user has specified the format is sentence;label;
# Reading with semicolon separator. The trailing semicolon will create an extra column.
# Using on_bad_lines='warn' to handle parsing errors.
df = pd.read_csv(input_file, sep=';', header=None, names=['sentence', 'label', 'extra'], engine='python', on_bad_lines='warn')

# Drop the extra column
if 'extra' in df.columns:
    df = df.drop(columns=['extra'])

# Apply the label mapping
df['label'] = df['label'].map(sentiment_label_mapping)

# Filter out rows to be discarded
df = df[df['label'] != '(Discard)']

# Save the preprocessed data
df.to_csv(output_file, index=False)

print(f"Preprocessing complete for {os.path.basename(input_file)}. Output saved to {output_file}")
print(df.head())
```

```python
import pandas as pd
import os

# Define the mapping for the sentiment labels
sentiment_label_mapping_tamil = {
    'Positive': 'Positive',
    'Negative': 'Negative',
    'unknown_state': 'Neutral',
    'Mixed_feelings': 'Neutral',
    'not-Tamil': '(Discard)'
}

# Define file paths
input_file = '/home/nibiru/Documents/projects/Text_try2/DravidianCodeMix/tamil_sentiment_full_train.csv'
output_dir = '/home/nibiru/Documents/projects/Text_try2/preprocessed_data'
output_filename = 'tamil_sentiment_full_train_preprocessed.csv'
output_file = os.path.join(output_dir, output_filename)

# Manual parsing, line by line
sentences = []
labels = []
with open(input_file, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        
        # Find the last tab and split there
        last_tab_index = line.rfind('\t')
        if last_tab_index != -1:
            sentence = line[:last_tab_index]
            label = line[last_tab_index+1:]
            sentences.append(sentence)
            labels.append(label)

# Create DataFrame
df = pd.DataFrame({
    'sentence': sentences,
    'label': labels
})

# Apply the label mapping
df['label'] = df['label'].map(sentiment_label_mapping_tamil)

# Filter out rows to be discarded
df = df[df['label'] != '(Discard)']

# Save the preprocessed data
df.to_csv(output_file, index=False)

print(f"Preprocessing complete for {os.path.basename(input_file)}. Output saved to {output_file}")
print(df.head())
```

```python
import pandas as pd
import os
import glob

# Define the directory containing the preprocessed files
preprocessed_dir = '/home/nibiru/Documents/projects/Text_try2/preprocessed_data'
output_dir = '/home/nibiru/Documents/projects/Text_try2/combined_data'

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Define the languages and their file prefixes
languages = {
    'kannada': 'kannada_',
    'malayalam': 'mal_',
    'tamil': 'tamil_'
}

for lang, prefix in languages.items():
    # Get a list of all preprocessed files for the current language
    file_pattern = os.path.join(preprocessed_dir, f"{prefix}*.csv")
    file_list = glob.glob(file_pattern)
    
    if not file_list:
        print(f"No files found for {lang}")
        continue
        
    # Create a list to hold the dataframes
    df_list = []
    
    # Read each file and append its dataframe to the list
    for file_path in file_list:
        df = pd.read_csv(file_path)
        df_list.append(df)
        
    # Concatenate all dataframes in the list
    combined_df = pd.concat(df_list, ignore_index=True)
    
    # Define the output file path
    output_file = os.path.join(output_dir, f"{lang}_combined.csv")
    
    # Save the combined dataframe to a new CSV file
    combined_df.to_csv(output_file, index=False)
    
    print(f"Combined {lang} files into {output_file}")
    print(f"Total rows for {lang}: {len(combined_df)}")
    print("-" * 30)
```

```python
import re
import emoji

def clean_text(text):
    # Lowercasing
    text = text.lower()
    # Removing URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    # Removing User Mentions
    text = re.sub(r'@\w+', '', text)
    # Convert Emojis to Text
    text = emoji.demojize(text, delimiters=(":", ":"))
    # Clean Hashtags
    text = re.sub(r'#(\w+)', r'\1', text)
    # Removing Special Characters
    text = re.sub(r'[^\w\s:]', '', text)
    # Removing Extra Whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
import pandas as pd
import os

combined_dir = '/home/nibiru/Documents/projects/Text_try2/combined_data'
cleaned_dir = '/home/nibiru/Documents/projects/Text_try2/cleaned_data'
os.makedirs(cleaned_dir, exist_ok=True)

combined_files = [f for f in os.listdir(combined_dir) if f.endswith('_combined.csv')]

for file_name in combined_files:
    file_path = os.path.join(combined_dir, file_name)
    df = pd.read_csv(file_path)
    
    # Drop rows where 'sentence' is NaN
    df.dropna(subset=['sentence'], inplace=True)
    
    df['sentence'] = df['sentence'].astype(str).apply(clean_text)
    
    # Define the output path
    output_file_path = os.path.join(cleaned_dir, file_name.replace('_combined.csv', '_cleaned.csv'))
    
    # Save the cleaned data
    df.to_csv(output_file_path, index=False)
    
    print(f"Cleaned and saved {file_name} to {output_file_path}")

print("All files have been cleaned and saved.")
```
